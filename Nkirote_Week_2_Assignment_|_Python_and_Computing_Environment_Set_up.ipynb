```python
#the fahrenheit threshold
thresh = 115

#function to convert celcius to farhenheit
def conversion(temp):
  farhn = (temp * 9/5) + 32

  #conditional statement to check whether the farhn temp is above or below threshold
  if farhn > thresh:
    print("temp inputed is above fahrenheit threshhold")
  else:
    print("temp inputed is below fahrenheit threshhold")

  print(f"temp in fahrenheit is: {farhn:.2f}")

#asking the user the number of temps they want to try
num_temp = int(input("how many temperatures will you input? "))

#loop for the requested number of temp attempts
for i in range(num_temp):
  temp = float(input(f"enter the temperature {i+1} in celcius: "))
  conversion(temp)
```
